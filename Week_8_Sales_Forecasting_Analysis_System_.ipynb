```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
import plotly.express as px
```

**1. Data Collection**

```python
data = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/train.csv")
```

```python
print("Shape:", data.shape)
print("\nColumn Names:")
print(data.columns)
data
```

**2. Data Cleaning & Preprocessing**

```python
print("Original Shape:", data.shape)

data = data[['Order Date', 'Region', 'Product ID', 'Sales', 'State']]

data = data.rename(columns={
    'Order Date': 'Date',
    'Product ID': 'Item',
    'State': 'Store'
})

print("New Shape:", data.shape)
data
```

```python
print(data['Date'].dtype)

data['Date'] = pd.to_datetime(data['Date'], errors='coerce')

print(data['Date'].dtype)

data
```

```python
data = data.drop_duplicates()

data = data.dropna()

data.reset_index(drop=True, inplace=True)

data.info()
```

```python
total_sales = data['Sales'].sum()
print("Total Sales:", total_sales)

average_sales = data['Sales'].mean()
print("Average Sales:", average_sales)

max_sales = data['Sales'].max()
min_sales = data['Sales'].min()
print("Max Sales:", max_sales)
print("Min Sales:", min_sales)
```

```python
fig = px.histogram(
    data,
    x="Sales",
    nbins=40,
    title="Sales Distribution",
    marginal="box",
)

fig.show()
```

**4. Sales Trend Analysis**

*Analyze sales over time*

```python
sales_over_time = data.groupby('Date')['Sales'].sum().reset_index()

fig = px.line(
    sales_over_time,
    x='Date',
    y='Sales',
    title='Sales Over Time'
)

fig.update_layout(
    xaxis_title='Date',
    yaxis_title='Total Sales',
    template='plotly_white'
)

fig.show()
```

*Identify seasonal trends*

```python
data['Year'] = data['Date'].dt.year
data['Month'] = data['Date'].dt.month
data['Month_Name'] = data['Date'].dt.strftime('%b')

data
```

```python
monthly_sales = data.groupby(['Year','Month','Month_Name'])['Sales'].sum().reset_index()

fig = px.line(
    monthly_sales,
    x="Month_Name",
    y="Sales",
    color="Year",
    markers=True,
    title="Monthly Sales Trend (Seasonal Analysis)"
)

fig.show()
```

*Detect peak sales periods*

```python
peak_months = monthly_sales.sort_values(by='Sales', ascending=False)

peak_months.head(10)

top_months = peak_months.head(10)
```

```python
top_months = peak_months.head(10)

fig = px.bar(
    top_months,
    x="Month_Name",
    y="Sales",
    color="Year",
    title="Top Peak Sales Months"
)

fig.show()
```

**5. Store/Product Analysis**

*Compare sales across stores*

```python
store_sales = data.groupby('Store')['Sales'].sum().sort_values(ascending=False).reset_index()

fig = px.bar(
    store_sales.head(10),
    x='Store',
    y='Sales',
    title='Top 10 Stores by Total Sales',
    color='Sales',
    color_continuous_scale='viridis'
)

fig.update_layout(
    xaxis_title='Store',
    yaxis_title='Total Sales',
    template='plotly_white'
)

fig.show()
```

*Identify best-selling products*

```python
product_sales = data.groupby('Item')['Sales'].sum().sort_values(ascending=False).reset_index()

fig = px.bar(
    product_sales.head(10),
    x='Item',
    y='Sales',
    title='Top 10 Products by Total Sales',
    color='Sales',
    color_continuous_scale='magma'
)

fig.update_layout(
    xaxis_title='Product',
    yaxis_title='Total Sales',
    template='plotly_white'
)

fig.show()
```

*Analyze low-performing products*

```python
product_sales = data.groupby('Item')['Sales'].sum().reset_index()

low_performing_products = product_sales.sort_values(by='Sales', ascending=True)

low_performing_products.head(10)
```

```python
bottom_products = low_performing_products.head(10)

fig = px.bar(
    bottom_products,
    x='Item',
    y='Sales',
    title='Lowest Performing Products (Bottom 10)',
    color='Sales',
    color_continuous_scale='Reds'
)

fig.update_layout(
    xaxis_title='Product ID',
    yaxis_title='Total Sales',
    xaxis_tickangle=45
)

fig.show()
```
